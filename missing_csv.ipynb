```python
import pandas as pd

# Путь к файлу
INPUT_FILE = "output_data/М-4_Дон_Москва_–_Воронеж_–_Ростов-на-Дону_–_Краснодар_–_Новороссийск/км_1443+500_а_д_М-4_Дон_Москва_–_Новороссийск.csv"
```

```python
import pandas as pd


def read_csv(path):
    # Читаем CSV без заголовков
    df = pd.read_csv(path, header=None, encoding="utf-8-sig")

    # Проверим число столбцов
    n_columns = df.shape[1]
    print(f"Столбцов в файле: {n_columns}")

    # Заголовки
    top_header = [
        "Дата",
        *["Общая интенсивность автомобилей"] * 3,
        *["Легковые (до 6 м)"] * 3,
        *["Малые груз. (6-9 м)"] * 3,
        *["Грузовые (9-13 м)"] * 3,
        *["Груз. большие (13-22 м)"] * 3,
        *["Автопоезда (22-30 м)"] * 3,
        *["Автобусы"] * 3,
        *["Мотоциклы"] * 3,
        *["Скорость, км/ч"] * 2,
        *["Загрузка, %"] * 2
    ]

    sub_header = [
        "",  # Дата
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Итого", "Прямое", "Обратное",
        "Прямое", "Обратное",
        "Прямое", "Обратное"
    ]

    # Проверка длины
    if len(top_header) != n_columns or len(sub_header) != n_columns:
        raise ValueError(f"❌ Размеры заголовков не совпадают с числом столбцов ({n_columns})\n"
                         f"top_header: {len(top_header)}, sub_header: {len(sub_header)}")

    # Назначение заголовков
    df.columns = pd.MultiIndex.from_arrays([top_header, sub_header])

    # Преобразуем дату
    date_col = [col for col in df.columns if col[0] == "Дата"][0]
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

    # Установим datetime-индекс
    df = df.set_index(date_col)
    df.index.name = "Дата"

    # Преобразуем все значения в float
    df = df.astype(float)

    # Удалим дубликаты по индексу (если есть)
    df = df[~df.index.duplicated(keep="first")]
    
    # Попробуем задать частоту (если данные регулярные)
    try:
        df = df.asfreq("H")
    except Exception as e:
        print(f"⚠️ Не удалось задать частоту: {e}")

    return df
```
